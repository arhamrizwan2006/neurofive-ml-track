```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import joblib
```

```python
pipeline = joblib.load('titanic_pipeline.pkl')
```

```python
print(pipeline)
```

```python
preprocessor = pipeline.named_steps['preprocessor']
```

```python
cat_pipeline = preprocessor.named_transformers_['cat']
```

```python
onehot = cat_pipeline.named_steps['onehot']
```

```python
cat_columns = ['Pclass', 'Sex', 'Embarked', 'Alone']
```

```python
for col, cats in zip(cat_columns, onehot.categories_):
    print(col, '->', cats)
```

```python
def build_input_row(pclass, sex, age, sibsp, parch, fare, embarked):
    family_size = sibsp + parch + 1
    alone = 1 if family_size == 1 else 0
    row = pd.DataFrame([{'Age': age, 'Fare': fare, 'SibSp': sibsp, 'Parch': parch, 
                         'FamilySize': family_size, 'Pclass': pclass, 'Sex': sex, 
                         'Embarked': embarked, 'Alone': alone}])
    return row
```

```python
test_row = build_input_row(pclass=1, sex='female', age=29, sibsp=0, parch=0, fare=100, embarked='S')
```

```python
print(test_row)
```

```python
prediction = pipeline.predict(test_row)
```

```python
probability = pipeline.predict_proba(test_row)
```

```python
print("Prediction (0=died, 1=survived):", prediction)
print("Probability [P(died), P(survived)]:", probability)
```
