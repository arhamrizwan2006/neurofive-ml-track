### R<sup>2</sup> :

R² tells us how good our model is at guessing house prices based on the features we gave it like size, quality, neighborhood etc. 
Our score was **0.83**, which means these features explain about **83%** of why prices go up or down between houses. The other 17% remaining is because of features we didn't include like the exact condition, age, market distance etc. 

This means our model understands the overall pattern of house prices well, but it may still make errors when predicting the price of individual houses.
